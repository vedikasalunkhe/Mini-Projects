```python
# ---------- 1. Imports ----------
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score

# ---------- 2. Data (replace with your own later) ----------
data = {
    "study_hours": [1, 2, 2.5, 3, 4, 4.5, 5, 6, 6.5, 7, 8, 9, 3.5, 5.5, 7.5, 2.8],
    "attendance":  [60, 65, 70, 70, 75, 80, 80, 85, 85, 90, 92, 95, 72, 82, 88, 68],
    "marks":       [35, 42, 47, 52, 58, 63, 66, 72, 75, 80, 86, 92, 55, 69, 83, 50],
}
df = pd.DataFrame(data)

# To use your own file instead, upload marks.csv to Colab (left sidebar, folder icon)
# with columns: study_hours, attendance, marks. Then uncomment the next line:
# df = pd.read_csv("marks.csv")

print("Data preview:")
print(df.head(), "\n")

# ---------- 3. Features and label ----------
X = df[["study_hours", "attendance"]]   # inputs
y = df["marks"]                         # what we predict

# ---------- 4. Train/test split ----------
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

# ---------- 5. Train the model ----------
model = LinearRegression()
model.fit(X_train, y_train)

# ---------- 6. Evaluate ----------
predictions = model.predict(X_test)
print("Average error (marks):", round(mean_absolute_error(y_test, predictions), 2))
print("R2 score (1.0 is perfect):", round(r2_score(y_test, predictions), 3), "\n")

# ---------- 7. Predict for a new case ----------
new_student = pd.DataFrame({"study_hours": [5.5], "attendance": [82]})
print("Predicted marks for 5.5 hrs study, 82% attendance:",
      round(model.predict(new_student)[0], 1), "\n")

# ---------- 8. Plot actual vs predicted ----------
plt.scatter(y_test, predictions)
plt.plot([y.min(), y.max()], [y.min(), y.max()], "r--")  # perfect-prediction line
plt.xlabel("Actual marks")
plt.ylabel("Predicted marks")
plt.title("Actual vs Predicted Marks")
plt.show()

# ---------- 9. Interactive prediction ----------
hours = float(input("Enter your study hours: "))
attend = float(input("Enter your attendance %: "))
my_input = pd.DataFrame({"study_hours": [hours], "attendance": [attend]})
print("Your predicted marks:", round(model.predict(my_input)[0], 1))
```
